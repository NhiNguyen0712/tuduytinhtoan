# Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểm tra xem nó có thể là độ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra màn hình.

```python
a,b,c=map(float,input().split())
if a+b<=c or b+c<=a or a+c<=b or a<=0 or b<=0 or c<=0:
    print("Khong la ba canh cua mot tam giac")
else:
    import math
    is_can= a==b or a==c or b==c
    is_vuong=a**2+b**2==c**2 or a**2+c**2==b**2 or b**2+c**2==a**2
    if a==b==c:
        print("Tam giac deu")
    elif is_can:
        print("Tam giac can")
    elif is_vuong:
        print("Tam giac vuong")
    elif is_can and is_vuong:
        print("Tam giac vuong can")
    else:
        print("Tam giac thuong")
P=(a+b+c)/2
S=math.sqrt(P*(P-a)*(P-b)*(P-c))
print(f"Chu vi tam giac la:{2*P}, Dien tich tam giac la:{S:.2f}")
```

# Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vào một điểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?

```python
a,b=map(int,input("toa do tam la:").split())
r=float(input("ban kinh la"))
x,y=map(float,input("toa do cua A la :").split())
import math
S=math.sqrt((a-x)**2+(b-y)**2)
if S==r:
    print("A thuoc duong tron")
else:
    print("A khong thuoc duong tron")
```
